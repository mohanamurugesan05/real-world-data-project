```python
#import required libraries
import pandas as pd
import matplotlib.pyplot as plt

```

```python
#load dataset
df=pd.read_csv("/content/amazon.csv.zip")

#display first 5 rows
print("First 5 Rows:")
display(df.head())
#statistical summary
print("\nStatistical Summary:")
display(df.describe())


```

```python
#missing value checking
print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing numeric values with the column mean
numeric_cols = df.select_dtypes(include='number').columns
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

# Fill missing text values with 'Unknown'
text_cols = df.select_dtypes(include='object').columns
for col in text_cols:
    df[col] = df[col].fillna('Unknown')
```

```python
#remove duplicates
print("\nDuplicate Rows:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
#descriptive analyses
# Clean and convert 'discounted_price' to numeric
df['discounted_price_numeric'] = df['discounted_price'].str.replace('₹', '', regex=False).str.replace(',', '', regex=False).astype(float)

top_categories = (
    df.groupby('category')['discounted_price_numeric']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop Categories by Sales (based on discounted price):")
display(top_categories)
```

```python
#correlation analyses
print("\nCorrelation Matrix:")
display(df.corr(numeric_only=True))
```

```python
#visualization (bar chart)
top_categories.plot(kind='bar')
plt.title('Top 10 Categories by Sales')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Scatter Plot: Discounted Price vs Rating

plt.figure(figsize=(8, 5))
plt.scatter(df['discounted_price'], df['rating'], alpha=0.6)

plt.title('Discounted Price vs Rating')
plt.xlabel('Discounted Price (₹)')
plt.ylabel('Rating')

plt.grid(True)
plt.show()
```

```python
#save cleaned data
df.to_csv('cleaned_retail_sales_dataset.csv', index=False)
print("\nCleaned dataset saved successfully!")
```
